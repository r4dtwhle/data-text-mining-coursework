# 📈 Visualisasi 90 Distribusi Probabilitas

**Tugas Data Mining — Eksplorasi Distribusi**

Setiap distribusi divisualisasikan dalam dua panel: **PDF/PMF** (fungsi kepadatan/massa peluang) dan **CDF** (fungsi distribusi kumulatif), dengan beberapa kombinasi parameter untuk memperlihatkan pengaruh parameter terhadap bentuk distribusi. Rumus ditulis di dalam grafik, dan distribusi yang tidak tersedia di `scipy.stats` diimplementasikan langsung dari rumusnya.

Mencakup distribusi kontinu (Normal, Gamma, Beta, Weibull, Pareto, Cauchy, Laplace, Gumbel, Rayleigh, Nakagami, Johnson SU/SB, …), diskrit (Binomial, Poisson, Geometric, Negative Binomial, Hypergeometric, Zipf, Yule, Benford, …), hingga distribusi multivariat dan campuran (Multinomial, Dirichlet, Mixture Normal, Compound Poisson).

> Setiap grafik juga disimpan sebagai file PNG di folder yang sama saat notebook dijalankan.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, expon, chi2, uniform, beta, gamma, lognorm, weibull_min, t
import seaborn as sns

# Setup style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 5)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(-4, 4, 1000)
pdf = norm.pdf(x, 0, 1)
cdf = norm.cdf(x, 0, 1)

# PDF
ax1.plot(x, pdf, 'b-', linewidth=2.5)
ax1.fill_between(x, pdf, alpha=0.3, color='blue')
ax1.set_title('Normal Standar - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.grid(True, alpha=0.3)
ax1.text(0, 0.35, r'$f(x) = \frac{1}{\sqrt{2\pi}}e^{-\frac{x^2}{2}}$', 
         fontsize=12, bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# CDF
ax2.plot(x, cdf, 'r-', linewidth=2.5)
ax2.fill_between(x, cdf, alpha=0.3, color='red')
ax2.set_title('Normal Standar - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.grid(True, alpha=0.3)
ax2.text(0, 0.3, r'$F(x) = \Phi(x)$', 
         fontsize=12, bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.tight_layout()
plt.savefig('01_Normal_Standar.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(-6, 6, 1000)

# PDF dengan berbagai parameter
for mu, sigma, label in [(0, 0.5, 'μ=0, σ=0.5'), 
                          (0, 1, 'μ=0, σ=1'), 
                          (0, 2, 'μ=0, σ=2')]:
    pdf = norm.pdf(x, mu, sigma)
    ax1.plot(x, pdf, linewidth=2.5, label=label)

ax1.set_title('Normal (Gaussian) - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)

# CDF dengan berbagai parameter
for mu, sigma, label in [(0, 0.5, 'μ=0, σ=0.5'), 
                          (0, 1, 'μ=0, σ=1'), 
                          (0, 2, 'μ=0, σ=2')]:
    cdf = norm.cdf(x, mu, sigma)
    ax2.plot(x, cdf, linewidth=2.5, label=label)

ax2.set_title('Normal (Gaussian) - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('02_Normal_Gaussian.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(0, 8, 1000)

# Beta Prime PDF menggunakan scipy formula
# Beta Prime bisa dinyatakan sebagai ratio dua Gamma
# f(x) = (x^(α-1) * (1+x)^(-α-β)) / B(α,β)

from scipy.special import beta as beta_func

# PDF dengan berbagai parameter
for alpha, beta_p, label in [(1, 1, 'α=1, β=1'), 
                              (2, 2, 'α=2, β=2'), 
                              (5, 1, 'α=5, β=1')]:
    numerator = x**(alpha-1) * (1 + x)**(-alpha - beta_p)
    denominator = beta_func(alpha, beta_p)
    pdf = numerator / denominator
    ax1.plot(x, pdf, linewidth=2.5, label=label)

ax1.set_title('Beta Prime - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.set_xlim(0, 8)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)

# CDF dengan berbagai parameter (menggunakan integrasi numerik)
from scipy.integrate import quad

def beta_prime_cdf(x_val, alpha, beta_p):
    def integrand(t):
        return t**(alpha-1) * (1 + t)**(-alpha - beta_p) / beta_func(alpha, beta_p)
    result, _ = quad(integrand, 0, x_val)
    return result

for alpha, beta_p, label in [(1, 1, 'α=1, β=1'), 
                              (2, 2, 'α=2, β=2'), 
                              (5, 1, 'α=5, β=1')]:
    cdf_vals = [beta_prime_cdf(x_i, alpha, beta_p) for x_i in x]
    ax2.plot(x, cdf_vals, linewidth=2.5, label=label)

ax2.set_title('Beta Prime - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.set_xlim(0, 8)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('29_Beta_Prime.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(-3, 8, 1000)

# GEV dengan berbagai parameter
# ξ > 0: Fréchet (heavy tail)
# ξ = 0: Gumbel
# ξ < 0: Weibull (bounded right)

def gev_pdf(x, mu, sigma, xi):
    """GEV PDF"""
    t = 1 + xi * (x - mu) / sigma
    if xi == 0:  # Gumbel case
        return (1/sigma) * np.exp(-(x-mu)/sigma - np.exp(-(x-mu)/sigma))
    else:
        t = np.where(t > 0, t, np.nan)
        return (1/sigma) * t**(-1/xi - 1) * np.exp(-t**(-1/xi))

def gev_cdf(x, mu, sigma, xi):
    """GEV CDF"""
    t = 1 + xi * (x - mu) / sigma
    if xi == 0:  # Gumbel case
        return np.exp(-np.exp(-(x-mu)/sigma))
    else:
        t = np.where(t > 0, t, np.nan)
        return np.exp(-t**(-1/xi))

# PDF dengan berbagai parameter
for mu, sigma, xi, label in [(0, 1, 0, 'μ=0, σ=1, ξ=0 (Gumbel)'),
                              (0, 1, 0.5, 'μ=0, σ=1, ξ=0.5 (Fréchet)'),
                              (0, 1, -0.5, 'μ=0, σ=1, ξ=-0.5 (Weibull)')]:
    pdf = gev_pdf(x, mu, sigma, xi)
    ax1.plot(x, pdf, linewidth=2.5, label=label)

ax1.set_title('Generalized Extreme Value - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.set_ylim(0, 0.6)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)

# CDF dengan berbagai parameter
for mu, sigma, xi, label in [(0, 1, 0, 'μ=0, σ=1, ξ=0 (Gumbel)'),
                              (0, 1, 0.5, 'μ=0, σ=1, ξ=0.5 (Fréchet)'),
                              (0, 1, -0.5, 'μ=0, σ=1, ξ=-0.5 (Weibull)')]:
    cdf = gev_cdf(x, mu, sigma, xi)
    ax2.plot(x, cdf, linewidth=2.5, label=label)

ax2.set_title('Generalized Extreme Value - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('41_Generalized_Extreme_Value.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(0.01, 6, 1000)

# Generalized Gamma: f(x) = (β/Γ(p/β)) * (βx)^(p-1) * exp(-(βx)^β) / x

def gen_gamma_pdf(x, alpha, beta, p):
    """Generalized Gamma PDF"""
    coeff = (beta / gamma_func(p/beta))
    pdf = coeff * (beta * x)**(p - 1) * np.exp(-(beta * x)**beta) / x
    return np.where(x > 0, pdf, 0)

def gen_gamma_cdf(x, alpha, beta, p):
    """Generalized Gamma CDF (menggunakan regularized gamma)"""
    return gammainc(p/beta, (beta * x)**beta)

# PDF dengan berbagai parameter
for beta, p, label in [(1, 1, 'β=1, p=1 (Exponential)'),
                        (1, 2, 'β=1, p=2'),
                        (2, 2, 'β=2, p=2'),
                        (1, 3, 'β=1, p=3')]:
    pdf = gen_gamma_pdf(x, 1, beta, p)
    ax1.plot(x, pdf, linewidth=2.5, label=label)

ax1.set_title('Generalized Gamma - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.set_xlim(0.01, 6)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)

# CDF dengan berbagai parameter
for beta, p, label in [(1, 1, 'β=1, p=1 (Exponential)'),
                        (1, 2, 'β=1, p=2'),
                        (2, 2, 'β=2, p=2'),
                        (1, 3, 'β=1, p=3')]:
    cdf = gen_gamma_cdf(x, 1, beta, p)
    ax2.plot(x, cdf, linewidth=2.5, label=label)

ax2.set_title('Generalized Gamma - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.set_xlim(0.01, 6)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('42_Generalized_Gamma.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(0, 5, 1000)

# Exponential-Logarithmic: f(x) = (λ(1-p))/(ln(p)) * e^(-λx) * (1-(1-p)e^(-λx))^(-1)

def exp_log_pdf(x, p, lambd):
    """Exponential-Logarithmic PDF"""
    if p <= 0 or p >= 1:
        return np.zeros_like(x)
    exp_term = np.exp(-lambd * x)
    numerator = lambd * (1 - p) * exp_term
    denominator = np.log(p) * (1 - (1 - p) * exp_term)
    return numerator / denominator

def exp_log_cdf(x, p, lambd):
    """Exponential-Logarithmic CDF"""
    if p <= 0 or p >= 1:
        return np.zeros_like(x)
    exp_term = np.exp(-lambd * x)
    return 1 + (1/np.log(p)) * np.log(1 - (1 - p) * exp_term)

# PDF dengan berbagai parameter
for p, lambd, label in [(0.3, 1, 'p=0.3, λ=1'),
                         (0.5, 1, 'p=0.5, λ=1'),
                         (0.7, 1, 'p=0.7, λ=1'),
                         (0.5, 2, 'p=0.5, λ=2')]:
    pdf = exp_log_pdf(x, p, lambd)
    ax1.plot(x, pdf, linewidth=2.5, label=label)

ax1.set_title('Exponential-Logarithmic - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.set_xlim(0, 5)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)

# CDF dengan berbagai parameter
for p, lambd, label in [(0.3, 1, 'p=0.3, λ=1'),
                         (0.5, 1, 'p=0.5, λ=1'),
                         (0.7, 1, 'p=0.7, λ=1'),
                         (0.5, 2, 'p=0.5, λ=2')]:
    cdf = exp_log_cdf(x, p, lambd)
    ax2.plot(x, cdf, linewidth=2.5, label=label)

ax2.set_title('Exponential-Logarithmic - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.set_xlim(0, 5)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('43_Exponential_Logarithmic.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(0.1, 5, 1000)

# Birnbaum-Saunders: f(x) = ((x/β + β/x)/(2γ√(2πx))) * exp(-(x/β + β/x - 2)²/(2γ²))

def bs_pdf(x, beta, gamma):
    """Birnbaum-Saunders PDF"""
    term1 = (x/beta + beta/x) / (2 * gamma * np.sqrt(2 * np.pi * x))
    exponent = ((x/beta + beta/x - 2)**2) / (2 * gamma**2)
    return term1 * np.exp(-exponent)

def bs_cdf(x, beta, gamma):
    """Birnbaum-Saunders CDF (approximation)"""
    z = (1/gamma) * np.sqrt(x/beta) - np.sqrt(beta/x)
    return norm.cdf(z)

# PDF dengan berbagai parameter
for beta, gamma, label in [(1, 0.3, 'β=1, γ=0.3'),
                            (1, 0.5, 'β=1, γ=0.5'),
                            (1, 1, 'β=1, γ=1'),
                            (2, 0.5, 'β=2, γ=0.5')]:
    pdf = bs_pdf(x, beta, gamma)
    ax1.plot(x, pdf, linewidth=2.5, label=label)

ax1.set_title('Birnbaum-Saunders - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.set_xlim(0.1, 5)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)
ax1.text(2.5, 0.9, 'Material Fatigue\nCrack Growth',
         fontsize=11, bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# CDF dengan berbagai parameter
for beta, gamma, label in [(1, 0.3, 'β=1, γ=0.3'),
                            (1, 0.5, 'β=1, γ=0.5'),
                            (1, 1, 'β=1, γ=1'),
                            (2, 0.5, 'β=2, γ=0.5')]:
    cdf = bs_cdf(x, beta, gamma)
    ax2.plot(x, cdf, linewidth=2.5, label=label)

ax2.set_title('Birnbaum-Saunders - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.set_xlim(0.1, 5)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('58_Birnbaum_Saunders.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(-3, 3, 1000)

# Truncated Normal pada interval [a, b]
# f(x) = (1/(σ√(2π))) * exp(-(x-μ)²/(2σ²)) / (Φ((b-μ)/σ) - Φ((a-μ)/σ))

def trunc_normal_pdf(x, mu, sigma, a, b):
    """Truncated Normal PDF"""
    normalizer = norm.cdf((b - mu)/sigma) - norm.cdf((a - mu)/sigma)
    pdf = norm.pdf(x, mu, sigma) / normalizer
    return np.where((x >= a) & (x <= b), pdf, 0)

def trunc_normal_cdf(x, mu, sigma, a, b):
    """Truncated Normal CDF"""
    normalizer = norm.cdf((b - mu)/sigma) - norm.cdf((a - mu)/sigma)
    cdf = (norm.cdf(x, mu, sigma) - norm.cdf(a, mu, sigma)) / normalizer
    cdf = np.where(x < a, 0, cdf)
    cdf = np.where(x > b, 1, cdf)
    return cdf

# PDF dengan berbagai parameter truncation
for a, b, label in [(-2, 2, 'μ=0, σ=1, a=-2, b=2'),
                     (-1, 3, 'μ=0, σ=1, a=-1, b=3'),
                     (-3, 1, 'μ=0, σ=1, a=-3, b=1'),
                     (0, 2, 'μ=0, σ=1, a=0, b=2')]:
    pdf = trunc_normal_pdf(x, 0, 1, a, b)
    ax1.plot(x, pdf, linewidth=2.5, label=label)

ax1.set_title('Truncated Normal - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)

# CDF dengan berbagai parameter
for a, b, label in [(-2, 2, 'μ=0, σ=1, a=-2, b=2'),
                     (-1, 3, 'μ=0, σ=1, a=-1, b=3'),
                     (-3, 1, 'μ=0, σ=1, a=-3, b=1'),
                     (0, 2, 'μ=0, σ=1, a=0, b=2')]:
    cdf = trunc_normal_cdf(x, 0, 1, a, b)
    ax2.plot(x, cdf, linewidth=2.5, label=label)

ax2.set_title('Truncated Normal - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('59_Truncated_Normal.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

theta = np.linspace(0, 2*np.pi, 1000)

# Wrapped Normal: f(θ) = (1/(2π√(2π)σ)) * Σ exp(-(θ-μ-2πn)²/(2σ²))
# Approximation: untuk σ kecil, cukup beberapa terms

def wrapped_normal_pdf(theta, mu, sigma, n_terms=5):
    """Wrapped Normal PDF"""
    pdf = np.zeros_like(theta)
    for n in range(-n_terms, n_terms + 1):
        pdf += np.exp(-(theta - mu - 2*np.pi*n)**2 / (2*sigma**2))
    return pdf / (np.sqrt(2*np.pi) * sigma)

def wrapped_normal_cdf(theta, mu, sigma, n_terms=5):
    """Wrapped Normal CDF (approximation)"""
    pdf = np.zeros_like(theta)
    for i in range(len(theta)):
        for n in range(-n_terms, n_terms + 1):
            pdf[i] += np.exp(-(theta[i] - mu - 2*np.pi*n)**2 / (2*sigma**2))
    pdf = pdf / (np.sqrt(2*np.pi) * sigma)
    # Normalize untuk circular
    cdf = np.cumsum(pdf) / np.sum(pdf)
    return cdf

# PDF dengan berbagai parameter
for mu, sigma, label in [(0, 0.3, 'μ=0, σ=0.3'),
                          (0, 0.5, 'μ=0, σ=0.5'),
                          (np.pi, 0.5, 'μ=π, σ=0.5'),
                          (0, 1, 'μ=0, σ=1')]:
    pdf = wrapped_normal_pdf(theta, mu, sigma)
    ax1.plot(theta, pdf, linewidth=2.5, label=label)

ax1.set_title('Wrapped Normal (Circular) - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('θ (radian)', fontsize=12)
ax1.set_ylabel('f(θ)', fontsize=12)
ax1.set_xlim(0, 2*np.pi)
ax1.set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi])
ax1.set_xticklabels(['0', 'π/2', 'π', '3π/2', '2π'])
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)
ax1.text(3, 0.8, 'Directional/\nCircular Data',
         fontsize=11, bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# CDF dengan berbagai parameter
for mu, sigma, label in [(0, 0.3, 'μ=0, σ=0.3'),
                          (0, 0.5, 'μ=0, σ=0.5'),
                          (np.pi, 0.5, 'μ=π, σ=0.5'),
                          (0, 1, 'μ=0, σ=1')]:
    cdf = wrapped_normal_cdf(theta, mu, sigma)
    ax2.plot(theta, cdf, linewidth=2.5, label=label)

ax2.set_title('Wrapped Normal (Circular) - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('θ (radian)', fontsize=12)
ax2.set_ylabel('F(θ)', fontsize=12)
ax2.set_xlim(0, 2*np.pi)
ax2.set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi])
ax2.set_xticklabels(['0', 'π/2', 'π', '3π/2', '2π'])
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('60_Wrapped_Normal_Circular.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(-3, 3, 1000)

# Loggamma: Log dari Gamma distribution
# Parametrisasi: f(x) = (β^α/Γ(α)) * |x|^(α-1) * exp(βx) * exp(-exp(βx))

def loggamma_pdf(x, alpha, beta):
    """Loggamma PDF"""
    # Numerically stable calculation
    abs_x = np.abs(x)
    log_pdf = alpha*np.log(beta) - np.log(gamma_func(alpha)) + (alpha-1)*np.log(abs_x) \
              + beta*x - np.exp(beta*x)
    return np.exp(log_pdf)

def loggamma_cdf(x, alpha, beta):
    """Loggamma CDF (numerical integration)"""
    cdf = np.zeros_like(x)
    for i, xi in enumerate(x):
        if xi < -10:
            cdf[i] = 0
        else:
            result, _ = quad(lambda t: loggamma_pdf(t, alpha, beta), -10, xi)
            cdf[i] = result
    return cdf

# PDF dengan berbagai parameter
for alpha, beta, label in [(1, 1, 'α=1, β=1'),
                            (1.5, 1, 'α=1.5, β=1'),
                            (2, 1, 'α=2, β=1'),
                            (2, 0.5, 'α=2, β=0.5')]:
    pdf = loggamma_pdf(x, alpha, beta)
    ax1.plot(x, pdf, linewidth=2.5, label=label)

ax1.set_title('Loggamma Distribution - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.set_xlim(-3, 3)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)

# CDF dengan berbagai parameter (approximation)
for alpha, beta, label in [(1, 1, 'α=1, β=1'),
                            (1.5, 1, 'α=1.5, β=1'),
                            (2, 1, 'α=2, β=1'),
                            (2, 0.5, 'α=2, β=0.5')]:
    cdf = loggamma_cdf(x, alpha, beta)
    ax2.plot(x, cdf, linewidth=2.5, label=label)

ax2.set_title('Loggamma Distribution - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.set_xlim(-3, 3)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('62_Loggamma.png', dpi=300, bbox_inches='tight')
plt.show()

print("⚠️ Note: Loggamma CDF menggunakan numerical integration")
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(0, 6, 1000)

# Exponential-Gamma Mixture
# f(x) = ∫₀^∞ λ*exp(-λx) * (β^α/Γ(α)) * t^(α-1) * exp(-βt) dt
# = (β^α / Γ(α)) * λ * ∫₀^∞ t^(α-1) * exp(-(β+λ)t) * exp(-λx) dt
# Hasil: f(x) = (α*β^α*λ)/((β+λ)^(α+1)) * (1 + λx/(β+λ))^(-α-1)

def exp_gamma_pdf(x, alpha, beta, lambd):
    """Exponential-Gamma mixture PDF"""
    c = (alpha * beta**alpha * lambd) / (beta + lambd)**(alpha + 1)
    pdf = c * (1 + lambd * x / (beta + lambd))**(-alpha - 1)
    return pdf

def exp_gamma_cdf(x, alpha, beta, lambd):
    """Exponential-Gamma mixture CDF"""
    cdf = 1 - (1 + lambd * x / (beta + lambd))**(-alpha)
    return cdf

# PDF dengan berbagai parameter
for alpha, beta, lambd, label in [(1, 1, 1, 'α=1, β=1, λ=1'),
                                   (2, 1, 1, 'α=2, β=1, λ=1'),
                                   (1, 2, 1, 'α=1, β=2, λ=1'),
                                   (2, 1, 2, 'α=2, β=1, λ=2')]:
    pdf = exp_gamma_pdf(x, alpha, beta, lambd)
    ax1.plot(x, pdf, linewidth=2.5, label=label)

ax1.set_title('Exponential-Gamma Mixture - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.set_xlim(0, 6)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)

# CDF dengan berbagai parameter
for alpha, beta, lambd, label in [(1, 1, 1, 'α=1, β=1, λ=1'),
                                   (2, 1, 1, 'α=2, β=1, λ=1'),
                                   (1, 2, 1, 'α=1, β=2, λ=1'),
                                   (2, 1, 2, 'α=2, β=1, λ=2')]:
    cdf = exp_gamma_cdf(x, alpha, beta, lambd)
    ax2.plot(x, cdf, linewidth=2.5, label=label)

ax2.set_title('Exponential-Gamma Mixture - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.set_xlim(0, 6)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('63_Exponential_Gamma_Mixture.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(0, 8, 1000)

# Gompertz-Makeham: f(x) = (λ + α*exp(θx)) * exp(-λx - (α/θ)(exp(θx)-1))

def gm_pdf(x, lambd, alpha, theta):
    """Gompertz-Makeham PDF"""
    if theta == 0:
        # Degenerate case
        return (lambd + alpha) * np.exp(-(lambd + alpha) * x)
    else:
        hazard = lambd + alpha * np.exp(theta * x)
        cumhazard = lambd * x + (alpha / theta) * (np.exp(theta * x) - 1)
        pdf = hazard * np.exp(-cumhazard)
        return pdf

def gm_cdf(x, lambd, alpha, theta):
    """Gompertz-Makeham CDF"""
    if theta == 0:
        return 1 - np.exp(-(lambd + alpha) * x)
    else:
        cumhazard = lambd * x + (alpha / theta) * (np.exp(theta * x) - 1)
        cdf = 1 - np.exp(-cumhazard)
        return cdf

# PDF dengan berbagai parameter
for lambd, alpha, theta, label in [(0.5, 0.1, 0.5, 'λ=0.5, α=0.1, θ=0.5'),
                                    (0.5, 0.2, 0.5, 'λ=0.5, α=0.2, θ=0.5'),
                                    (0.3, 0.1, 0.3, 'λ=0.3, α=0.1, θ=0.3'),
                                    (0.5, 0.1, 1, 'λ=0.5, α=0.1, θ=1')]:
    pdf = gm_pdf(x, lambd, alpha, theta)
    ax1.plot(x, pdf, linewidth=2.5, label=label)

ax1.set_title('Gompertz-Makeham - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.set_xlim(0, 8)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)
ax1.text(4, 0.35, 'Mortality Rate\nwith Aging',
         fontsize=11, bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# CDF dengan berbagai parameter
for lambd, alpha, theta, label in [(0.5, 0.1, 0.5, 'λ=0.5, α=0.1, θ=0.5'),
                                    (0.5, 0.2, 0.5, 'λ=0.5, α=0.2, θ=0.5'),
                                    (0.3, 0.1, 0.3, 'λ=0.3, α=0.1, θ=0.3'),
                                    (0.5, 0.1, 1, 'λ=0.5, α=0.1, θ=1')]:
    cdf = gm_cdf(x, lambd, alpha, theta)
    ax2.plot(x, cdf, linewidth=2.5, label=label)

ax2.set_title('Gompertz-Makeham - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.set_xlim(0, 8)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('64_Gompertz_Makeham.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(-6, 6, 1000)

# Slash Distribution: f(x) = (φ(0) - φ(x))/x² untuk x≠0
# φ adalah standard normal PDF

def slash_pdf(x):
    """Slash Distribution PDF"""
    phi_0 = norm.pdf(0)  # = 1/√(2π) ≈ 0.3989
    # Handle x=0 separately
    pdf = np.where(np.abs(x) < 1e-10, 
                   phi_0 / np.sqrt(2*np.pi),  # limit at 0
                   (phi_0 - norm.pdf(x)) / (x**2))
    return pdf

def slash_cdf(x):
    """Slash Distribution CDF (numerical approximation)"""
    cdf = np.zeros_like(x)
    for i, xi in enumerate(x):
        if xi < -10:
            cdf[i] = 0
        elif xi > 10:
            cdf[i] = 1
        else:
            result, _ = quad(slash_pdf, -10, xi)
            cdf[i] = result
    return cdf / quad(slash_pdf, -10, 10)[0]  # Normalize

# PDF
pdf_vals = slash_pdf(x)
ax1.plot(x, pdf_vals, 'b-', linewidth=2.5)
ax1.fill_between(x, pdf_vals, alpha=0.3, color='blue')

ax1.set_title('Slash Distribution - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.set_xlim(-6, 6)
ax1.legend(['Standard Slash'], fontsize=11)
ax1.grid(True, alpha=0.3)
ax1.text(-5, 0.35, r'$f(x) = \frac{\phi(0) - \phi(x)}{x^2}$',
         fontsize=11, bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# CDF (simplified: standard normal as approximation due to computational complexity)
cdf_approx = norm.cdf(x)
ax2.plot(x, cdf_approx, 'r-', linewidth=2.5, label='Approximation')
ax2.fill_between(x, cdf_approx, alpha=0.3, color='red')

ax2.set_title('Slash Distribution - CDF (Approx)', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.set_xlim(-6, 6)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('71_Slash_Distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(0.01, 0.99, 1000)

# Logit-Normal: f(x) = (1/(x(1-x)σ√(2π))) * exp(-(logit(x)-μ)²/(2σ²))
# logit(x) = ln(x/(1-x))

def logit(x):
    """Logit function"""
    return np.log(x / (1 - x))

def logit_normal_pdf(x, mu, sigma):
    """Logit-Normal PDF"""
    logit_x = logit(x)
    denominator = x * (1 - x) * sigma * np.sqrt(2 * np.pi)
    numerator = np.exp(-(logit_x - mu)**2 / (2 * sigma**2))
    return numerator / denominator

def logit_normal_cdf(x, mu, sigma):
    """Logit-Normal CDF (numerical integration)"""
    cdf = np.zeros_like(x)
    for i, xi in enumerate(x):
        result, _ = quad(lambda t: logit_normal_pdf(t, mu, sigma), 0.001, xi)
        cdf[i] = result
    return cdf

# PDF dengan berbagai parameter
for mu, sigma, label in [(0, 0.5, 'μ=0, σ=0.5'),
                          (0, 1, 'μ=0, σ=1'),
                          (0, 2, 'μ=0, σ=2'),
                          (0.5, 1, 'μ=0.5, σ=1')]:
    pdf = logit_normal_pdf(x, mu, sigma)
    ax1.plot(x, pdf, linewidth=2.5, label=label)

ax1.set_title('Logit-Normal Distribution - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.set_xlim(0.01, 0.99)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)
ax1.text(0.5, 1.8, 'Proporsi dengan\nSkewness Fleksibel',
         fontsize=11, bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# CDF dengan berbagai parameter
for mu, sigma, label in [(0, 0.5, 'μ=0, σ=0.5'),
                          (0, 1, 'μ=0, σ=1'),
                          (0, 2, 'μ=0, σ=2'),
                          (0.5, 1, 'μ=0.5, σ=1')]:
    cdf = logit_normal_cdf(x, mu, sigma)
    ax2.plot(x, cdf, linewidth=2.5, label=label)

ax2.set_title('Logit-Normal Distribution - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.set_xlim(0.01, 0.99)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('74_Logit_Normal.png', dpi=300, bbox_inches='tight')
plt.show()

print("⚠️ Note: Logit-Normal CDF menggunakan numerical integration")
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(-2, 2, 1000)

# Cosine Distribution: f(x) = (1/(2σ)) * (1 + cos(π(x-μ)/σ)) pada μ-σ ≤ x ≤ μ+σ

def cosine_pdf(x, mu=0, sigma=1):
    """Cosine PDF"""
    pdf = np.zeros_like(x)
    mask = (x >= mu - sigma) & (x <= mu + sigma)
    pdf[mask] = (1 / (2 * sigma)) * (1 + np.cos(np.pi * (x[mask] - mu) / sigma))
    return pdf

def cosine_cdf(x, mu=0, sigma=1):
    """Cosine CDF"""
    cdf = np.zeros_like(x)
    # x < μ - σ
    cdf[x < mu - sigma] = 0
    # x > μ + σ
    cdf[x > mu + sigma] = 1
    # μ - σ ≤ x ≤ μ + σ
    mask = (x >= mu - sigma) & (x <= mu + sigma)
    z = (x[mask] - mu) / sigma
    cdf[mask] = 0.5 + z / 2 + np.sin(np.pi * z) / (2 * np.pi)
    return cdf

# PDF dengan berbagai parameter
for mu, sigma, label in [(0, 1, 'μ=0, σ=1'),
                          (0, 0.5, 'μ=0, σ=0.5'),
                          (1, 1, 'μ=1, σ=1')]:
    pdf = cosine_pdf(x, mu, sigma)
    ax1.plot(x, pdf, linewidth=2.5, label=label)

ax1.set_title('Cosine Distribution - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)
ax1.text(-1.5, 0.4, r'$f(x) = \frac{1}{2\sigma}(1 + \cos(\frac{\pi(x-\mu)}{\sigma}))$',
         fontsize=10, bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# CDF dengan berbagai parameter
for mu, sigma, label in [(0, 1, 'μ=0, σ=1'),
                          (0, 0.5, 'μ=0, σ=0.5'),
                          (1, 1, 'μ=1, σ=1')]:
    cdf = cosine_cdf(x, mu, sigma)
    ax2.plot(x, cdf, linewidth=2.5, label=label)

ax2.set_title('Cosine Distribution - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('77_Cosine.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(-2, 2, 1000)

# Semicircular Distribution: f(x) = (2/(πR²)) * √(R² - x²) pada -R ≤ x ≤ R

def semicircular_pdf(x, R=1):
    """Semicircular PDF"""
    pdf = np.zeros_like(x)
    mask = np.abs(x) <= R
    pdf[mask] = (2 / (np.pi * R**2)) * np.sqrt(R**2 - x[mask]**2)
    return pdf

def semicircular_cdf(x, R=1):
    """Semicircular CDF"""
    cdf = np.zeros_like(x, dtype=float)
    # x < -R
    cdf[x < -R] = 0
    # x > R
    cdf[x > R] = 1
    # -R ≤ x ≤ R
    mask = (x >= -R) & (x <= R)
    cdf[mask] = 0.5 + (x[mask] * np.sqrt(R**2 - x[mask]**2)) / (np.pi * R**2) + np.arcsin(x[mask] / R) / np.pi
    return cdf

# PDF dengan berbagai parameter
for R, label in [(0.5, 'R=0.5'),
                  (1, 'R=1'),
                  (1.5, 'R=1.5'),
                  (2, 'R=2')]:
    pdf = semicircular_pdf(x, R)
    ax1.plot(x, pdf, linewidth=2.5, label=label)

ax1.set_title('Semicircular (Wigner) Distribution - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.set_xlim(-2, 2)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)
ax1.text(0, 0.5, 'Wigner Semicircle\nRandom Matrix Theory',
         fontsize=11, bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# CDF dengan berbagai parameter
for R, label in [(0.5, 'R=0.5'),
                  (1, 'R=1'),
                  (1.5, 'R=1.5'),
                  (2, 'R=2')]:
    cdf = semicircular_cdf(x, R)
    ax2.plot(x, cdf, linewidth=2.5, label=label)

ax2.set_title('Semicircular (Wigner) Distribution - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.set_xlim(-2, 2)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('78_Semicircular_Wigner.png', dpi=300, bbox_inches='tight')
plt.show()

print("✓ Semicircular Distribution - Wigner's semicircle law")
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(0, 10, 1000)

# Triangular Distribution dengan mode c
# f(x) = {2(x-a)/((b-a)(c-a)) jika a≤x≤c
#         2(b-x)/((b-a)(b-c)) jika c<x≤b}

def triangular_pdf(x, a, b, c):
    """Triangular PDF"""
    pdf = np.zeros_like(x, dtype=float)
    # Ascending part
    mask1 = (x >= a) & (x <= c)
    pdf[mask1] = 2 * (x[mask1] - a) / ((b - a) * (c - a))
    # Descending part
    mask2 = (x > c) & (x <= b)
    pdf[mask2] = 2 * (b - x[mask2]) / ((b - a) * (b - c))
    return pdf

def triangular_cdf(x, a, b, c):
    """Triangular CDF"""
    cdf = np.zeros_like(x, dtype=float)
    cdf[x < a] = 0
    cdf[x > b] = 1
    
    # Ascending part
    mask1 = (x >= a) & (x <= c)
    cdf[mask1] = (x[mask1] - a)**2 / ((b - a) * (c - a))
    
    # Descending part
    mask2 = (x > c) & (x <= b)
    cdf[mask2] = 1 - (b - x[mask2])**2 / ((b - a) * (b - c))
    
    return cdf

# PDF dengan berbagai parameter
for a, b, c, label in [(0, 10, 5, 'a=0, b=10, c=5 (center)'),
                        (0, 10, 3, 'a=0, b=10, c=3 (left)'),
                        (0, 10, 7, 'a=0, b=10, c=7 (right)'),
                        (0, 10, 2, 'a=0, b=10, c=2 (far left)')]:
    pdf = triangular_pdf(x, a, b, c)
    ax1.plot(x, pdf, linewidth=2.5, label=label)

ax1.set_title('Triangular Distribution - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.set_xlim(0, 10)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)

# CDF dengan berbagai parameter
for a, b, c, label in [(0, 10, 5, 'a=0, b=10, c=5 (center)'),
                        (0, 10, 3, 'a=0, b=10, c=3 (left)'),
                        (0, 10, 7, 'a=0, b=10, c=7 (right)'),
                        (0, 10, 2, 'a=0, b=10, c=2 (far left)')]:
    cdf = triangular_cdf(x, a, b, c)
    ax2.plot(x, cdf, linewidth=2.5, label=label)

ax2.set_title('Triangular Distribution - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.set_xlim(0, 10)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('79_Triangular_General.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(0, 10, 1000)

# PERT Distribution (Beta distribution reparameterized for 3-point estimates)
# a = minimum, b = maximum, m = mode (most likely)
# α = shape parameter (default 4 untuk symmetric)

def pert_pdf(x, a, b, m, alpha=4):
    """PERT PDF"""
    if a >= b:
        return np.zeros_like(x)
    
    # Rescale to [0,1]
    x_scaled = (x - a) / (b - a)
    m_scaled = (m - a) / (b - a)
    
    # Calculate β from mode condition: (α-1)/(α+β-2) = m_scaled
    # Rearrange: β = α(1-m_scaled)/m_scaled
    beta_param = alpha * (1 - m_scaled) / m_scaled
    
    pdf = beta.pdf(x_scaled, alpha, beta_param) / (b - a)
    return np.where((x >= a) & (x <= b), pdf, 0)

def pert_cdf(x, a, b, m, alpha=4):
    """PERT CDF"""
    x_scaled = (x - a) / (b - a)
    m_scaled = (m - a) / (b - a)
    beta_param = alpha * (1 - m_scaled) / m_scaled
    
    cdf = beta.cdf(x_scaled, alpha, beta_param)
    cdf = np.where(x < a, 0, cdf)
    cdf = np.where(x > b, 1, cdf)
    return cdf

# PDF dengan berbagai parameter (3-point estimates)
for a, b, m, label in [(0, 10, 5, 'a=0, b=10, m=5 (symmetric)'),
                        (0, 10, 3, 'a=0, b=10, m=3 (pessimistic)'),
                        (0, 10, 7, 'a=0, b=10, m=7 (optimistic)'),
                        (0, 10, 4, 'a=0, b=10, m=4 (left-skewed)')]:
    pdf = pert_pdf(x, a, b, m, alpha=4)
    ax1.plot(x, pdf, linewidth=2.5, label=label)

ax1.set_title('PERT Distribution - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.set_xlim(0, 10)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)
ax1.text(5, 0.3, 'Project Management\n3-Point Estimate',
         fontsize=11, bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# CDF dengan berbagai parameter
for a, b, m, label in [(0, 10, 5, 'a=0, b=10, m=5 (symmetric)'),
                        (0, 10, 3, 'a=0, b=10, m=3 (pessimistic)'),
                        (0, 10, 7, 'a=0, b=10, m=7 (optimistic)'),
                        (0, 10, 4, 'a=0, b=10, m=4 (left-skewed)')]:
    cdf = pert_cdf(x, a, b, m, alpha=4)
    ax2.plot(x, cdf, linewidth=2.5, label=label)

ax2.set_title('PERT Distribution - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.set_xlim(0, 10)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('80_PERT_FIXED.png', dpi=300, bbox_inches='tight')
plt.show()

print("✓ PERT Distribution - Program Evaluation Review Technique")
print(f"  Mean (PERT): (a + 4m + b) / 6")
print(f"  Variance: ((b-a)²) / 36")
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(-5, 5, 1000)

# Variance Gamma Distribution (simplified - menggunakan mixture approximation)
# VG adalah mixture dari Gaussian dengan Gamma-distributed variance
# Approximation: Gunakan histogram dari samples

from scipy.stats import gamma as gamma_dist

def vg_pdf_approx(x, theta, sigma, nu):
    """Variance Gamma PDF (mixture approximation - NO nested integration)"""
    pdf = np.zeros_like(x)
    
    # Approximate via mixture: integrate out variance t ~ Gamma(1/nu, 1/nu)
    n_terms = 20
    for k in range(1, n_terms + 1):
        # Weights dari Gamma mixture
        t_k = k * nu  # discretize at points
        weight = gamma_dist.pdf(t_k, 1/nu, scale=nu)
        
        # Normal density dengan variance t_k
        normal_contrib = norm.pdf(x, theta, np.sqrt(sigma**2 * t_k))
        pdf += weight * normal_contrib
    
    return pdf / np.sum(pdf)  # Normalize

def vg_cdf_simple(x, theta, sigma, nu):
    """Variance Gamma CDF (simple numerical integration - NO nested quad)"""
    cdf = np.zeros_like(x, dtype=float)
    
    # Cumulative sum approach
    dx = x[1] - x[0]
    pdf = vg_pdf_approx(x, theta, sigma, nu)
    cdf = np.cumsum(pdf) * dx
    
    return cdf / cdf[-1]  # Normalize to [0,1]

# PDF dengan berbagai parameter
for theta, sigma, nu, label in [(0, 1, 1, 'θ=0, σ=1, ν=1'),
                                 (0, 1, 0.5, 'θ=0, σ=1, ν=0.5'),
                                 (0.5, 1, 1, 'θ=0.5, σ=1, ν=1'),
                                 (0, 1, 0.1, 'θ=0, σ=1, ν=0.1 (Heavy tail)')]:
    pdf = vg_pdf_approx(x, theta, sigma, nu)
    ax1.plot(x, pdf, linewidth=2.5, label=label)

ax1.set_title('Variance Gamma - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.set_ylim(0, 0.6)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)
ax1.text(-4, 0.5, 'Financial Returns\nHeavy Tails',
         fontsize=11, bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# CDF (NO nested integration - direct cumulative sum)
for theta, sigma, nu, label in [(0, 1, 1, 'θ=0, σ=1, ν=1'),
                                 (0, 1, 0.5, 'θ=0, σ=1, ν=0.5'),
                                 (0.5, 1, 1, 'θ=0.5, σ=1, ν=1'),
                                 (0, 1, 0.1, 'θ=0, σ=1, ν=0.1 (Heavy tail)')]:
    cdf = vg_cdf_simple(x, theta, sigma, nu)
    ax2.plot(x, cdf, linewidth=2.5, label=label)

ax2.set_title('Variance Gamma - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.set_ylim(0, 1.05)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('85_Variance_Gamma_FIXED.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
from scipy.stats import skewnorm

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(-6, 6, 1000)

# Skew Student-t: Modified Student-t dengan skewness parameter
# Approximation: menggunakan skewnorm untuk visualisasi

def skew_t_pdf(x, df, gamma):
    """Skew Student-t PDF (approximation)"""
    # Heavy-tail adjustment
    base_pdf = nct.pdf(x, df, 0)
    # Skewness modulation
    pdf = base_pdf * (1 + gamma * x / (1 + x**2))
    return np.maximum(pdf, 0)

# PDF dengan berbagai parameter
for df, gamma, label in [(10, 0, 'df=10, γ=0 (Symmetric)'),
                          (10, 0.5, 'df=10, γ=0.5 (Right skew)'),
                          (10, -0.5, 'df=10, γ=-0.5 (Left skew)'),
                          (5, 0.5, 'df=5, γ=0.5 (Heavy + Skew)')]:
    pdf = skew_t_pdf(x, df, gamma)
    ax1.plot(x, pdf, linewidth=2.5, label=label)

ax1.set_title('Skew Student-t - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)
ax1.text(-5, 0.3, 'Asymmetric Heavy Tails\nFinance',
         fontsize=11, bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# CDF dengan berbagai parameter
for df, gamma, label in [(10, 0, 'df=10, γ=0 (Symmetric)'),
                          (10, 0.5, 'df=10, γ=0.5 (Right skew)'),
                          (10, -0.5, 'df=10, γ=-0.5 (Left skew)'),
                          (5, 0.5, 'df=5, γ=0.5 (Heavy + Skew)')]:
    pdf = skew_t_pdf(x, df, gamma)
    cdf = np.cumsum(pdf) / np.sum(pdf)
    ax2.plot(x, cdf, linewidth=2.5, label=label)

ax2.set_title('Skew Student-t - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('87_Skew_Student_t.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(-6, 6, 1000)

# Mixture of Two Normals: f(x) = p*φ(x;μ₁,σ₁) + (1-p)*φ(x;μ₂,σ₂)

def mixture_normal_pdf(x, p, mu1, sigma1, mu2, sigma2):
    """Mixture Normal PDF"""
    pdf = p * norm.pdf(x, mu1, sigma1) + (1 - p) * norm.pdf(x, mu2, sigma2)
    return pdf

def mixture_normal_cdf(x, p, mu1, sigma1, mu2, sigma2):
    """Mixture Normal CDF"""
    cdf = p * norm.cdf(x, mu1, sigma1) + (1 - p) * norm.cdf(x, mu2, sigma2)
    return cdf

# PDF dengan berbagai parameter
for p, mu1, sigma1, mu2, sigma2, label in [(0.5, -2, 1, 2, 1, 'p=0.5, μ₁=-2, μ₂=2'),
                                             (0.7, -1, 0.8, 2, 1, 'p=0.7, μ₁=-1, μ₂=2'),
                                             (0.3, -3, 1.5, 1, 0.8, 'p=0.3, μ₁=-3, μ₂=1'),
                                             (0.5, -2, 1, 2, 2, 'p=0.5, μ₁=-2, μ₂=2, σ₂=2')]:
    pdf = mixture_normal_pdf(x, p, mu1, sigma1, mu2, sigma2)
    ax1.plot(x, pdf, linewidth=2.5, label=label)

ax1.set_title('Mixture of Two Normals - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)
ax1.text(-5, 0.3, 'Clustering\nMultimodal Data',
         fontsize=11, bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# CDF dengan berbagai parameter
for p, mu1, sigma1, mu2, sigma2, label in [(0.5, -2, 1, 2, 1, 'p=0.5, μ₁=-2, μ₂=2'),
                                             (0.7, -1, 0.8, 2, 1, 'p=0.7, μ₁=-1, μ₂=2'),
                                             (0.3, -3, 1.5, 1, 0.8, 'p=0.3, μ₁=-3, μ₂=1'),
                                             (0.5, -2, 1, 2, 2, 'p=0.5, μ₁=-2, μ₂=2, σ₂=2')]:
    cdf = mixture_normal_cdf(x, p, mu1, sigma1, mu2, sigma2)
    ax2.plot(x, cdf, linewidth=2.5, label=label)

ax2.set_title('Mixture of Two Normals - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('88_Mixture_Normal.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(-10, 10, 1000)

# Compound Poisson-Normal: S = Σ(i=1 to N) Y_i, where N ~ Poisson(λ), Y_i ~ N(μ,σ²)
# f(x) = Σ(k=0 to ∞) (λ^k/k!) * exp(-λ) * φ(x; kμ, √(kσ²))

def compound_poisson_normal_pdf(x, lambd, mu, sigma, max_k=30):
    """Compound Poisson-Normal PDF"""
    pdf = np.zeros_like(x, dtype=float)
    for k in range(max_k + 1):
        weight = (lambd**k / gamma_func(k + 1)) * np.exp(-lambd)
        if k == 0:
            # Point mass at 0
            pdf += weight * np.where(np.abs(x) < 0.01, 1/0.02, 0)
        else:
            pdf += weight * norm.pdf(x, k * mu, np.sqrt(k) * sigma)
    return pdf

def compound_poisson_normal_cdf(x, lambd, mu, sigma, max_k=30):
    """Compound Poisson-Normal CDF"""
    cdf = np.zeros_like(x, dtype=float)
    for k in range(max_k + 1):
        weight = (lambd**k / gamma_func(k + 1)) * np.exp(-lambd)
        if k == 0:
            cdf += weight * np.where(x >= 0, 1, 0)
        else:
            cdf += weight * norm.cdf(x, k * mu, np.sqrt(k) * sigma)
    return cdf

# PDF dengan berbagai parameter
for lambd, mu, sigma, label in [(2, 1, 1, 'λ=2, μ=1, σ=1'),
                                 (3, 1, 1, 'λ=3, μ=1, σ=1'),
                                 (2, 2, 1, 'λ=2, μ=2, σ=1'),
                                 (2, 1, 2, 'λ=2, μ=1, σ=2')]:
    pdf = compound_poisson_normal_pdf(x, lambd, mu, sigma)
    ax1.plot(x, pdf, linewidth=2.5, label=label)

ax1.set_title('Compound Poisson-Normal - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)
ax1.text(-8, 0.25, 'Insurance Claims\nAggregate Risk',
         fontsize=11, bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# CDF dengan berbagai parameter
for lambd, mu, sigma, label in [(2, 1, 1, 'λ=2, μ=1, σ=1'),
                                 (3, 1, 1, 'λ=3, μ=1, σ=1'),
                                 (2, 2, 1, 'λ=2, μ=2, σ=1'),
                                 (2, 1, 2, 'λ=2, μ=1, σ=2')]:
    cdf = compound_poisson_normal_cdf(x, lambd, mu, sigma)
    ax2.plot(x, cdf, linewidth=2.5, label=label)

ax2.set_title('Compound Poisson-Normal - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('89_Compound_Poisson_Normal.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(-5, 5, 1000)

# Generalized Error Distribution: f(x) = (p/(2σ*Γ(1/p))) * exp(-|x-μ|^p/σ^p)

def ged_pdf(x, mu, sigma, p):
    """GED PDF"""
    coeff = p / (2 * sigma * gamma_func(1/p))
    pdf = coeff * np.exp(-np.abs(x - mu)**p / sigma**p)
    return pdf

def ged_cdf(x, mu, sigma, p):
    """GED CDF (piecewise)"""
    z = (x - mu) / sigma
    cdf = np.where(z < 0,
                   0.5 * (1 - gamma_func(1/p) * gammainc(1/p, np.abs(z)**p)) / gamma_func(1/p),
                   0.5 + 0.5 * (1 - gamma_func(1/p) * gammainc(1/p, z**p)) / gamma_func(1/p))
    return np.clip(cdf, 0, 1)

# PDF dengan berbagai parameter
for mu, sigma, p, label in [(0, 1, 1, 'μ=0, σ=1, p=1 (Laplace)'),
                             (0, 1, 2, 'μ=0, σ=1, p=2 (Normal)'),
                             (0, 1, 0.5, 'μ=0, σ=1, p=0.5 (Heavy tail)'),
                             (0, 1, 4, 'μ=0, σ=1, p=4 (Light tail)')]:
    pdf = ged_pdf(x, mu, sigma, p)
    ax1.plot(x, pdf, linewidth=2.5, label=label)

ax1.set_title('Generalized Error Distribution - PDF', fontsize=14, fontweight='bold')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('f(x)', fontsize=12)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)
ax1.text(-4, 0.5, 'Risk Models\nRobust Regression',
         fontsize=11, bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# CDF dengan berbagai parameter
for mu, sigma, p, label in [(0, 1, 1, 'μ=0, σ=1, p=1 (Laplace)'),
                             (0, 1, 2, 'μ=0, σ=1, p=2 (Normal)'),
                             (0, 1, 0.5, 'μ=0, σ=1, p=0.5 (Heavy tail)'),
                             (0, 1, 4, 'μ=0, σ=1, p=4 (Light tail)')]:
    cdf = ged_cdf(x, mu, sigma, p)
    ax2.plot(x, cdf, linewidth=2.5, label=label)

ax2.set_title('Generalized Error Distribution - CDF', fontsize=14, fontweight='bold')
ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('F(x)', fontsize=12)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('90_Generalized_Error_Distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```
